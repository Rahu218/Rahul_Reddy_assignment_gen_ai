```python
import pandas as pd
import numpy as np
```

```python
# Import csv file
df = pd.read_csv('Sales Dataset/Amazon Sale Report.csv')
```

```python
# Convert MM-DD-YY → datetime
df['Date'] = pd.to_datetime(df['Date'], format='%m-%d-%y')
df['Date'] = df['Date'].dt.strftime('%Y-%m-%d')
```

```python
# Remove columns with names "index" and "Unnamed: 22"
df = df.loc[:, ~df.columns.isin(['index', 'Unnamed: 22'])]
```
